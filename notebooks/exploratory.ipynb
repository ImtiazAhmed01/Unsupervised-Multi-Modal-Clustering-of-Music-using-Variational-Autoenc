```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# 1. Setup Path to import from src
sys.path.append(os.path.abspath('../'))
# Now we can import your custom function
# from src.dataset import load_data 

# 2. Load Data directly for visualization
# Change path to '../data/features_30_sec.csv' if running from notebooks/ folder
df = pd.read_csv('../data/features_30_sec.csv')

# 3. Set visualization style
sns.set_theme(style="whitegrid")

# --- VISUALIZATION 1: TEMPO ---
plt.figure(figsize=(12, 6))
sns.histplot(data=df, x='tempo', hue='label', element='step', palette='tab10', common_norm=False)
plt.title('How Tempo Varies Across Genres', fontsize=15)
plt.xlabel('Tempo (BPM)')
plt.show()

# --- VISUALIZATION 2: SPECTRAL CENTROID (Brightness) ---
plt.figure(figsize=(12, 6))
sns.histplot(data=df, x='spectral_centroid_mean', hue='label', element='step', palette='tab10', common_norm=False)
plt.title('Spectral Centroid (Sound "Brightness") by Genre', fontsize=15)
plt.xlabel('Spectral Centroid Mean')
plt.show()

# --- VISUALIZATION 3: CHROMA (Harmonic Content) ---
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='label', y='chroma_stft_mean', palette='tab10')
plt.title('Chroma STFT Mean (Harmonics) across Genres', fontsize=15)
plt.xticks(rotation=45)
plt.show()
```
